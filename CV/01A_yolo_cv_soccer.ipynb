```python
import matplotlib.pyplot as plt
import cv2
import random

# Select 15 random training images
sample_images = random.sample(train_images, 15)

# Create 3 × 5 subplot
fig, axes = plt.subplots(3, 5, figsize=(20, 11))

for ax, image_path in zip(axes.flat, sample_images):

    # Read image
    image = cv2.imread(str(image_path))

    # Convert BGR → RGB for Matplotlib
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Display
    ax.imshow(image)
    ax.set_title(image_path.name, fontsize=9)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
model = YOLO("yolo26n.pt")
```

```python
pprint(model.names)
```

```python
print("Person:", model.names[0])
print("Sports ball:", model.names[32])
```

```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA version:", torch.version.cuda)
```

```python
device = "cuda:0" if torch.cuda.is_available() else "cpu"

print("Using device:", device)
```

```python
image_path = random.choice(val_images)

results = model.predict(
    source=str(image_path),
    imgsz=1280,
    conf=0.25,
    classes=[0, 32],
    device=device,
    verbose=False
)

result = results[0]

annotated = result.plot(
    font_size=8,
    line_width=2
)

plt.figure(figsize=(16, 9))
plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
