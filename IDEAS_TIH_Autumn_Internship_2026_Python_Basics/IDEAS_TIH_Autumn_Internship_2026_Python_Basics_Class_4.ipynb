```python
# Abstraction:   Hiding the implementational details of a class and showing only the essential features to the user
# Encapsulation: Wrapping data and functions into a single unit(object). Basically it is data + method

class Car:
  def __init__(self, number):
    self.acc = False
    self.clutch = False
    self.brake = False
    self.carNumber = number
  def startCar(self):
    self.clutch = True
    self.acc = True
    print(f"In car {self.carNumber}, clutch turned {self.clutch} and accelerator turned {self.acc} so start the car")
  def stopCar(self):
    self.brake = True
    self.acc = False
    print(f"In car {self.carNumber}, Brake turned {self.brake} and accelerator turned {self.acc} so stop the car")

car1 = Car(1)
car2 = Car(2)

car1.startCar()
car2.startCar()

car1.stopCar()
car2.stopCar()
```

```python
# Inheritance:

class Employee: # Parent class
  def __init__(self, name, id):
    self.name = name
    self.id = id

  def showEmployeeDetails(self):
    print(f"The name of employee {self.id} is {self.name} ")

class Programmer(Employee): # child class
  def showLanguage(self, lang):
    print(f"The Programmer {self.name} works using {lang}")

# e1 = Employee("Archit", 101)
# e1.showEmployeeDetails()

p1 = Programmer("Bipasha", 102)
p1.showEmployeeDetails()
p1.showLanguage("Java")
```

```python
class A:
  def feature1(self):
    print("This is feature 1 of class A")
  def feature2(self):
    print("This is feature 2 of class A")

class B(A): # single-level inheritance
  def feature3(self):
    print("This is feature 3 of class B")
  def feature4(self):
    print("This is feature 4 of class B")

class C(B): # multi-level inheritance
  def feature5(self):
    print("This is feature 5 of class C")
  def feature6(self):
    print("This is feature 6 of class C")

if __name__ == "__main__":
  a = A()
  b = B()
  c = C()

  print(f"{"*"*40}Using class A{"*"*40}")
  a.feature1()
  a.feature2()

  print(f"{"*"*40}Using class B{"*"*40}")
  b.feature1()
  b.feature2()
  b.feature3()
  b.feature4()

  print(f"{"*"*40}Using class C{"*"*40}")
  c.feature5()
  c.feature4()
  c.feature1()
```

```python
'''
1D - [Axis 0]
2D - [Axis 0 (Row), Axis 1 (Column)]
'''


X = [[1,2,3], [4,5,6], [7,8,9]]
arr=np.array(X)
print(arr, arr.shape)
print(f"Summation of rows = {arr.sum(axis=0)}") # column is constant, addition occurs between different row elements
print(f"Summation of columns = {arr.sum(axis=1)}") # row is constant, addition occurs between different column elements
print(f"Transpose of a matrix = \n{arr.T}")
print(f"Sum of all the elements of a single row = {arr[2].sum()}")
print(f"Sum of all the elements of a single column = {arr[:,2].sum()}")
print(f"Trace of the matrix (summation of the primary diagonal elements) = {np.trace(X)}")
print(f"Determinant of the matrix = {np.linalg.det(X)}")
```

```python
arr_flat = arr.flatten()
print("The flattened version of the matrix  =", arr_flat)
print(arr.ravel())
```
