```python
# WAPP to print all the prime numbers within a range [lb,ub] using functions

lb = int(input("Enter the lower bound = "))
ub = int(input("Enter the upper bound = "))

def isPrime(n):
  for i in range(2,n):
    if n%i==0:
      return False
  return True

for i in range(lb, ub+1):
  if isPrime(i):
    print(f"{i} is a PRIME NUMBER")
```

```python
# WAPP to checkk if a number is an Armstrong Number
# 153 = 1^3 + 5^3 + 3^3

def isArmstrong(n):
  l = len(str(n))
  temp = n
  sum = 0
  while(temp>0):
    rem = temp%10
    sum = sum + pow(rem,l)
    temp=temp//10
  if sum == n:
    print(f"{n} is an ARMSTRONG NUMBER")
  else:
    print(f"{n} is not an ARMSTRONG NUMBER")

isArmstrong(int(input("Enter a number to check if it is an Armstrong Number = ")))
```

```python
# RECURSION : A function calls itself

# Write a program to calculate thr factorial of a number using recursion
# 5! = 5*4*3*2*1 = 5 * 4!
# 4! = 4*3*2*1 = 4*3!
# 3! = 3*2*1
# n! = n*(n-1)!

def factorial(x):
  # base case:
  if(x==0 or x==1):
    return 1
  # recursive case:
  else:
    return x * factorial(x-1)

n = int(input("Enter a number = "))
print(f"The factorial of {n} is {factorial(n)}")

'''
factorial(5)
  is (5==0 or 5==1) ? NO
  return 5 * factorial(4) = 5 * 24

factorial(4)
  is (4==0 or 4==1) ? NO
  return 4 * factorial(3) = 4 * 6  = 24

factorial(3)
  is (3==0 or 3==1) ? NO
  return 3 * factorial(2) = 3 * 2 = 6

factorial(2)
  is (2==0 or 2==1) ? NO
  return 2 * factorial(1) = 2 * 1 = 2

factorial(1)
  is (1==0 or 1==1) ? TRUE
  return 1
'''

fact = 1
for i in range(1,n+1):
  fact = fact * i
print(fact)
```
