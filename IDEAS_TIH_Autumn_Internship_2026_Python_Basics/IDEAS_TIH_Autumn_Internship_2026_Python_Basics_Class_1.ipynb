```python
# variable : A named memory location. To store some value.. Can the value be changed later on ? YES!
a=12
print(a)
a=21
print(a)

# integer variables : will contain integer values
myInteger = 16754
print(myInteger, type(myInteger)) # type(): function that returns the datatype of the variable used

# float variables : will contain decimal values
pi = 3.14
print(pi, type(pi))

# string variables : will contain sentences or words etc..
name = "Harry Potter"
print(name, type(name))

# boolean variables : True or False
isThisClassInteresting = True
print(isThisClassInteresting, type(isThisClassInteresting))
```

```python
# Type Casting: Conversion of one datatype to another

#implicit typecasting
print(myInteger, type(myInteger))
myInteger = myInteger + pi
print(myInteger, type(myInteger))

#explicit typecasting / coercion
myInteger = int(myInteger)
print(myInteger, type(myInteger))

myInteger = str(myInteger)
print(myInteger, type(myInteger))

myInteger = bool(myInteger)
print(myInteger, type(myInteger))

print(int(isThisClassInteresting))
```

```python
# Basic Operations:
numerator = 11
denominator = 2
print("numerator + denominator = {numerator+denominator}") # f -> formatted strings
print(f"numerator + denominator = {numerator+denominator}") # Addition
print(f"numerator - denominator = {numerator-denominator}") # Subtraction
print(f"numerator * denominator = {numerator*denominator}") # Multiplication
print(f"numerator / denominator = {numerator/denominator}") # float division
print(f"numerator // denominator = {numerator//denominator}") # integer division
print(int(2.9999))
print(f"numerator % denominator = {numerator%denominator}")
print(f"numerator ** denominator = {numerator**denominator}")

demon = 0
try:
  print(numerator/demon)
except Exception as emotionalDamage:
  print(emotionalDamage)

print("My name is Voldemort")
```

```python
# Conditional Statements:

'''
if condition1:
  statement1
elif condition2: (elif means else if)
  statement2
else:
  statement3
'''


physics = 60
maths = 60

if(physics >= 85 and maths >=85): # and here is a logical operator and the result is TRUE if both the conditions are TRUE
  print("Pen + Diary")
elif(physics >= 85 or maths >=85 ): # or here is a logical operator and the result is TRUE if either condition or both the conditions are TRUE
  print("Pen")
else:
  print("FLYING CHAPPAL RECEIVED")
```
