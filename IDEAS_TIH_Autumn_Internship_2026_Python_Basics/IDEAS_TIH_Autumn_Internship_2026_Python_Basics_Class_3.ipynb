```python
# Lists revisited:

# append : add at the end of a list
myList = [1,2,3,4,5,6,7,8,9,7]
print(f"original list = {myList}")
myList.append(25)
print(f"modified list = {myList}")
# remove an element from the list
myList.remove(7) # remove(n) : removes the first occurence of n from the list
print(f"modified list after removing an element from the list =  {myList}")
#pop
myList.pop(7)
print(f"modified list after popping an element from the list =  {myList}")
#by default pop() removes the last element if no index is specified
myList.pop()
print(f"modified list after popping an element from the list =  {myList}")
#insert an element at some specific index
myList.insert(1,100)
print(f"modified list after inserting an element into the list =  {myList}")

# Lists are mutable: we can update a value at some given index
myList[3] = 10
myList[5] = 20
print(f"modified list after updating an element of the list =  {myList}")
print("*"*90)
fruits = ['apple', 'banana', 'cherry', 'orange', 'mango']
# create a fruit basket that will only carry fruits having an 'o' in it's name.
fruitBasket = []
for i in fruits:
  if 'o' in i:
    fruitBasket.append(i)
print(f"My fruit basket = {fruitBasket}")

# list comprehension:
fruitBasket_lc = [i for i in fruits if 'o' in i]
print(f"My fruit basket using list comprehension = {fruitBasket}")
print("*"*90)
# list input using list comprehension
# approach 1:
'''
# myListInput = [int(x) for x in input().split()]
# print(myListInput)
'''
#approach 2:
'''
lst = [int(input("Enter a no.")) for _ in range (5)]
print(lst)'''
```

```python
# Dictionaries: key-value pair
hogwarts = {"Gryffindor":'Hermione', "Ravenclaw":'Luna', "Slytherin": "Draco", 'Hufflepuff':'Cedric', 'Gryffindor':'Harry'}
print(hogwarts)
# print(hogwarts[0]) : An element of a Dictionary can be accessed only with the key value and not any index value
print(hogwarts['Slytherin'])
print(hogwarts.keys()) # keys(): collects all the keys associated in the dictionary
print(hogwarts.values()) # values(): collects all the values associated in the dictionary
print(hogwarts.items()) # items(): collects all the key-value pairs associated in the dictionary
# Dictonaries are mutable
hogwarts['Gryffindor'] = "Ron"
print(hogwarts)

for key, value in hogwarts.items():
  print(f"key = {key} and value = {value}" )
```

```python
# __init__(): Constructor of a class
# All Classes have an __init__(), which is always executed when the class is initiated .. i.e. when an object is created

class Student:
  def __init__(self, fullName):
    self.name = fullName
  def printName(self):
    print(f"self value = {self}")
    return self.name
s1 = Student("Miguel Diaz")
s2 = Student("Robby Lawrence")

print(f"{"*"*20} DEALING WITH object s1 {"*"*20}")
print(f"object value = {s1}")
name1 = s1.printName()
print(name1)
print(f"{"*"*20} DEALING WITH object s2 {"*"*20}")
print(f"object value = {s1}")
name2 = s2.printName()
print(name2)
# self: the self keyword is a reference to the current object
# self is also used to access variables that belong to a class
```

```python
# constructors are of 3 types : 1. default constructor 2. non-parameterized 3. paramterized constructor


# example of non-parameterised constructor
class Student:
  def __init__(self):
    self.sub1 = 'Physics'
    self.sub2 = 'Mathematics'
  def printName(self, fullName):
    print(f"The student {fullName} has taken the courses {self.sub1} and {self.sub2}")

s1 = Student()
s1.printName("Miguel")

s2= Student()
s2.printName("Robby")
```

```python
# example of non-parameterised constructor
'''
Class and Instance variables:

Class variable:
A variable which has a fixed value even if the objects change

Instance variables:
The values of the variables sub1, sub2, fullName, totalMarks may change with different objects. ie different objects may have different values for these variables.
thus, they are instance variables.
'''

'''
Method : is  a function that is defined in a class
'''

class Student:
  institute = "IDEAS-TIH"
  def __init__(self, sub1, sub2, name):
    self.sub1 = sub1
    self.sub2 = sub2
    self.fullName = name

  @staticmethod
  def printMyWish(myName):
    print(f"I {myName}, wish this class ends soon !!!!!!!!!!")

  def printMyWishSecond(self):
    print(f"I am {self.fullName} and I wish to have a cup of tea after this class is over !")

  def printMarks(self, totalMarks):
    print(f"The student {self.fullName} of {Student.institute} has taken the courses {self.sub1} and {self.sub2} and scored a total of = {totalMarks}")
    self.printMyWishSecond()
    Student.printMyWish(self.fullName)


s1 = Student('Mathematics', 'Physics', "Miguel")
s1.printMarks(85)

# s2= Student('Mathematics', 'Computer Science', "Robby")
# s2.printMarks(86)
```
