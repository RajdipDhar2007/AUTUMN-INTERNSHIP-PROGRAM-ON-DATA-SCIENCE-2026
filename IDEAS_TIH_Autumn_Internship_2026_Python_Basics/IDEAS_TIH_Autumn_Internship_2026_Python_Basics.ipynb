```python
# WAPP to checkk if a number is an Armstrong Number
# 153 = 1^3 + 5^3 + 3^3

def isArmstrong(n):
  l = len(str(n))
  temp = n
  sum = 0
  while(temp>0):
    rem = temp%10
    sum = sum + pow(rem,l)
    temp=temp//10
  if sum == n:
    print(f"{n} is an ARMSTRONG NUMBER")
  else:
    print(f"{n} is not an ARMSTRONG NUMBER")

isArmstrong(int(input("Enter a number to check if it is an Armstrong Number = ")))
```

```python
# RECURSION : A function calls itself

# Write a program to calculate thr factorial of a number using recursion
# 5! = 5*4*3*2*1 = 5 * 4!
# 4! = 4*3*2*1 = 4*3!
# 3! = 3*2*1
# n! = n*(n-1)!

def factorial(x):
  # base case:
  if(x==0 or x==1):
    return 1
  # recursive case:
  else:
    return x * factorial(x-1)

n = int(input("Enter a number = "))
print(f"The factorial of {n} is {factorial(n)}")

'''
factorial(5)
  is (5==0 or 5==1) ? NO
  return 5 * factorial(4) = 5 * 24

factorial(4)
  is (4==0 or 4==1) ? NO
  return 4 * factorial(3) = 4 * 6  = 24

factorial(3)
  is (3==0 or 3==1) ? NO
  return 3 * factorial(2) = 3 * 2 = 6

factorial(2)
  is (2==0 or 2==1) ? NO
  return 2 * factorial(1) = 2 * 1 = 2

factorial(1)
  is (1==0 or 1==1) ? TRUE
  return 1
'''

fact = 1
for i in range(1,n+1):
  fact = fact * i
print(fact)
```

```python
# example of non-parameterised constructor
'''
Class and Instance variables:

Class variable:
A variable which has a fixed value even if the objects change

Instance variables:
The values of the variables sub1, sub2, fullName, totalMarks may change with different objects. ie different objects may have different values for these variables.
thus, they are instance variables.
'''

'''
Method : is  a function that is defined in a class
'''

class Student:
  institute = "IDEAS-TIH"
  def __init__(self, sub1, sub2, name):
    self.sub1 = sub1
    self.sub2 = sub2
    self.fullName = name

  @staticmethod
  def printMyWish(myName):
    print(f"I {myName}, wish this class ends soon !!!!!!!!!!")

  def printMyWishSecond(self):
    print(f"I am {self.fullName} and I wish to have a cup of tea after this class is over !")

  def printMarks(self, totalMarks):
    print(f"The student {self.fullName} of {Student.institute} has taken the courses {self.sub1} and {self.sub2} and scored a total of = {totalMarks}")
    self.printMyWishSecond()
    Student.printMyWish(self.fullName)


s1 = Student('Mathematics', 'Physics', "Miguel")
s1.printMarks(85)

# s2= Student('Mathematics', 'Computer Science', "Robby")
# s2.printMarks(86)
```

```python
# Abstraction:   Hiding the implementational details of a class and showing only the essential features to the user
# Encapsulation: Wrapping data and functions into a single unit(object). Basically it is data + method

class Car:
  def __init__(self, number):
    self.acc = False
    self.clutch = False
    self.brake = False
    self.carNumber = number
  def startCar(self):
    self.clutch = True
    self.acc = True
    print(f"In car {self.carNumber}, clutch turned {self.clutch} and accelerator turned {self.acc} so start the car")
  def stopCar(self):
    self.brake = True
    self.acc = False
    print(f"In car {self.carNumber}, Brake turned {self.brake} and accelerator turned {self.acc} so stop the car")

car1 = Car(1)
car2 = Car(2)

car1.startCar()
car2.startCar()

car1.stopCar()
car2.stopCar()
```

```python
# Inheritance:

class Employee: # Parent class
  def __init__(self, name, id):
    self.name = name
    self.id = id

  def showEmployeeDetails(self):
    print(f"The name of employee {self.id} is {self.name} ")

class Programmer(Employee): # child class
  def showLanguage(self, lang):
    print(f"The Programmer {self.name} works using {lang}")

# e1 = Employee("Archit", 101)
# e1.showEmployeeDetails()

p1 = Programmer("Bipasha", 102)
p1.showEmployeeDetails()
p1.showLanguage("Java")
```

```python
class A:
  def feature1(self):
    print("This is feature 1 of class A")
  def feature2(self):
    print("This is feature 2 of class A")

class B(A): # single-level inheritance
  def feature3(self):
    print("This is feature 3 of class B")
  def feature4(self):
    print("This is feature 4 of class B")

class C(B): # multi-level inheritance
  def feature5(self):
    print("This is feature 5 of class C")
  def feature6(self):
    print("This is feature 6 of class C")

if __name__ == "__main__":
  a = A()
  b = B()
  c = C()

  print(f"{"*"*40}Using class A{"*"*40}")
  a.feature1()
  a.feature2()

  print(f"{"*"*40}Using class B{"*"*40}")
  b.feature1()
  b.feature2()
  b.feature3()
  b.feature4()

  print(f"{"*"*40}Using class C{"*"*40}")
  c.feature5()
  c.feature4()
  c.feature1()
```
